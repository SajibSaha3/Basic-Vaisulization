```python
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
x = [10,15,20,25]
y = [5,10,15,20]
plt.figure(figsize=(4,3))
plt.xlabel("value of x")
plt.ylabel("value of y")
plt.title("representation of x and y")
# plt.axis("off")
plt.plot(x,y,"o", color ="r")
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.xlabel("value of x")
plt.ylabel("value of y")
plt.title("representation of x and y")
# plt.axis("off")
plt.plot(x,y,"--", color ="r")
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.xlabel("value of x")
plt.ylabel("value of y")
plt.title("representation of x and y")
# plt.axis("off")
plt.plot(x,y,"^k--", color ="r")
plt.show()
```

```python
x = [10,15,2,25]
y = [5,10,5,20]

plt.figure(figsize=(4,3))
plt.xlabel("value of x")
plt.ylabel("value of y")
plt.title("representation of x and y")
# plt.axis("off")
plt.plot(x,y, color ="r")
```

```python
plt.figure(figsize=(4,3))
plt.xlabel("value of x")
plt.ylabel("value of y")
plt.title("representation of x and y")
# plt.axis("off")
plt.plot(x,y,"o", color ="r", marker="*")
plt.savefig('my_plot.png')
plt.show()
```

```python
# scatter plot 
x = [10,15,2,25,18,21]
y = [5,10,5,20,15,17]
z = ["red", "blue","red", "blue","red", "blue"]
plt.figure(figsize=(4,3))
plt.xlabel("value of x")
plt.ylabel("value of y")
plt.title("representation of x and y")
plt.scatter(x,y, color =z, s= 100)
```

```python
# scatter plot 
x = [10,15,2,25,18,21]
y = [5,10,5,20,15,17]
z = ["red", "blue","red", "blue","red", "blue"]
plt.figure(figsize=(4,3))
plt.xlabel("value of x")
plt.ylabel("value of y")
plt.title("representation of x and y")
plt.scatter(x,y, color =z, s= 100)
plt.show()
```

```python
# scatter plot with seaborn 
z = ["apple", "berry","strawberry", "orange","mango", "jack"]
plt.figure(figsize=(4,3))
plt.xlabel("value of x")
plt.ylabel("value of y")
plt.title("fruit value identificartion")
sns.scatterplot(x=x,y=y, hue =z, s=100)
plt.show()
```
